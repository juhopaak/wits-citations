```python
import pandas as pd
```

```python
df = pd.read_csv( "../data/LitLangMed_annotated.csv", dtype={43:str, 44:str, 45:str, 46:str}, keep_default_na=False, na_values=[''])
df_llm_annotated = pd.read_csv('LLM_annotated_data/coded_haiku_v3_final.csv')
```

```python
df_llm_annotated['Function of ref'] = df_llm_annotated['Function of ref'].str.replace('THEORY FRAMEWORK', 'THEORY')
```

```python
df['Location in thesis'] = df['Location in thesis'].fillna(df_llm_annotated['Location in thesis'])
df['Function of ref'] = df['Function of ref'].fillna(df_llm_annotated['Function of ref'])
df['Quality of ref'] = df['Quality of ref'].fillna(df_llm_annotated['Quality of ref'])
df['Number occurance'] = df['Number occurance'].fillna(df_llm_annotated['Number of occurrences'])
```

```python
df.to_csv('../data/LitLangMed_llm_annotated.csv', index=False)
df.to_excel('LitLangMed_llm_annotated.xlsx', index=False)
```
