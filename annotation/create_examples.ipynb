```python
import pandas as pd
```

```python
df = pd.read_excel('LitLangMed_preannotated.xlsx')
```

```python
categories = set()

for cat in df['Location in thesis'].value_counts().keys():

    for c in cat.split(';'):

        categories.add(c.strip().replace(" / ", "/"))

categories
```
